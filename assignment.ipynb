# Linear Algebra and Probability

```python
# Importing the modules
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

### 1. Load the dataset and handle any missing data by replacing null values with the mean value of the respective column.

```python
# Loading the dataset
data=pd.read_csv("winequality-red.csv")
# Viewing the first 10 rows of the dataset
data.head(10)
```

```python
# Displaying information about the dataset
print(data.info())
```

```python
# Checking for null values 
data.isna().sum()
```

***No null values are present in any rows of the dataset.***

```python
# Replacing any null value with the mean value of the respective column
data['fixed acidity']=data['fixed acidity'].fillna(data['fixed acidity'].mean())
data['volatile acidity']=data['volatile acidity'].fillna(data['volatile acidity'].mean())
data['citric acid']=data['citric acid'].fillna(data['citric acid'].mean())
data['residual sugar']=data['residual sugar'].fillna(data['residual sugar'].mean())
data['chlorides']=data['chlorides'].fillna(data['chlorides'].mean())
data['free sulfur dioxide']=data['free sulfur dioxide'].fillna(data['free sulfur dioxide'].mean())
data['total sulfur dioxide']=data['total sulfur dioxide'].fillna(data['total sulfur dioxide'].mean())
data['density']=data['density'].fillna(data['density'].mean())
data['pH']=data['pH'].fillna(data['pH'].mean())
data['sulphates']=data['sulphates'].fillna(data['sulphates'].mean())
data['alcohol']=data['alcohol'].fillna(data['alcohol'].mean())
data['quality']=data['quality'].fillna(data['quality'].mean())
```

```python
# Checking for null values 
data.isna().sum()
```

### 2. Extract the following columns as vectors: alcohol,citric acid.

```python
# Extracting the alcohol column as vector
alcohol_vector=data['alcohol']
# Printing the alcohol vector
print(f"Alcohol vector: {alcohol_vector.values}")
```

```python
# Extracting the citric acid column as vector
citric_acid_vector=data['citric acid']
# Printing the citric acid vector
print(f"Citric acid vector: {citric_acid_vector.values}")
```

### 3. Select two features (e.g., alcohol and density) from the dataset and calculate the covariance matrix using np.cov(X.T), where X is the feature matrix consisting of the selected columns.

```python
# Creating the feature matrix
X=data[['alcohol','density']]
# Displaying the feature matrix
print("Feature matrix:")
print(X)
```

```python
# Calculating the covariance matrix
covariance_matrix=np.cov(X.T)
# Printing the covariance matrix
print("Covariance matrix: ")
print(covariance_matrix)
```

### 4. Perform eigen decomposition on the covariance matrix you computed in question 3. Identify and interpret the results:Identify the top 2 eigenvalues of the covariance matrix,Identify the corresponding eigenvectors.

```python
# Calculating the eigen values and eigen vectors
eigen_values, eigen_vectors=np.linalg.eig(covariance_matrix)
```

```python
# Printing the eigen values
print("Eigen values:")
print(eigen_values)
```

```python
# Printing the eigen vectors
print("Eigen vectors:")
print(eigen_vectors)
```

```python
# Printing the eigen value with the corresponding eigen vector
print(f"Eigen value: {eigen_values[0]}, Eigen vector: {eigen_vectors[0]}")
print(f"Eigen value: {eigen_values[1]}, Eigen vector: {eigen_vectors[1]}")
```

### 5. Which wine quality is most common in the dataset? How can you interpret the distribution of wine quality scores?

```python
# Identifying the most common wine quality
most_common_wine_quality=data['quality'].mode()
print(f"Most common wine quality: {most_common_wine_quality[0]}")
```

```python
# Plotting a histogram to view the distribution of wine quality
fig, ax=plt.subplots(figsize=(5,5))
sns.histplot(
    x="quality",
    data=data,
    discrete=True
)
fig.suptitle("Most common wine quality", fontsize=18)
ax.set_title("Distribution of wine quality", fontsize=14)
ax.set_xlabel("Quality of wine", fontsize=12)
ax.set_ylabel("Count", fontsize=12)
ax.set_yticks(range(0,900,100))
ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)
plt.show()
```

```python
# Calculating the mean of the quality column
quality_mean=data['quality'].mean()
print(f"Mean of the quality column: {quality_mean}")
```

```python
# Calculating the median of the quality column
quality_median=data['quality'].median()
print(f"Median of the quality column: {quality_median}")
```

```python
# Calculating the skewness of the quality column
quality_skew=data['quality'].skew()
print(f"Skewness of the quality column: {quality_skew}")
```

***Since skewness of the quality column is 0.22 the data is moderately and positively skewed.***
